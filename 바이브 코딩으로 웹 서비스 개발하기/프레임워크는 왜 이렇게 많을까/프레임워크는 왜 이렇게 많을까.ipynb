앞 레슨에서는 HTML이 내용과 구조를, CSS가 표현을, JavaScript가 동작을 담당한다는 것을 살펴보았습니다. 웹서비스를 만들다 보면 React, Vue 같은 도구도 만나게 됩니다. 이 도구들은 어떤 일을 덜어주며, 내 서비스에는 무엇이 필요할까요?

이 레슨에서는 서비스의 조건을 AI에게 설명하고, 후보의 장단점과 선택 근거를 검토합니다. 결과는 다음 순서로 정리합니다.

1. **선택 근거**: 내 조건에 맞춰 후보 2～3개를 비교하고, 채택하거나 제외한 이유를 적습니다.
2. **PRD**(Product Requirements Document, 제품 요구사항 문서): 만들 서비스의 요구사항을 문서로 정리합니다.
3. **목업 데이터 필요 여부**: 내 서비스에서 목업 데이터가 필요한 곳과 그 이유를 정리합니다.

자신의 서비스로 선택 근거와 PRD를 작성한 뒤, 수제 디저트 쇼핑몰 예제로 목업 화면을 만들고 확인하는 과정까지 따라갑니다.

### 프레임워크가 제공하는 것

같은 모양의 카드를 여러 번 보여주거나, 검색어가 바뀔 때 목록을 다시 보여주는 일은 웹 화면에서 자주 반복됩니다. **프레임워크(framework)** 는 이런 작업에 필요한 기능과 프로그램을 구성하는 규칙을 제공합니다.

필요한 기능을 내 코드에서 가져다 쓰는 도구는 **라이브러리** 라고 합니다. 프레임워크는 기능뿐 아니라 프로그램의 실행 흐름과 작성 규칙도 제공합니다. React는 공식적으로 화면을 만드는 UI 라이브러리이지만, 이 레슨에서는 화면 개발을 돕는 도구로서 프레임워크와 함께 비교합니다.

화면을 만드는 것 외에 주소별로 화면을 연결하거나 서버에서 코드를 실행해야 할 수도 있습니다. Next.js는 React에 이런 기능을 더하는 **메타프레임워크** 입니다. 이름의 분류를 외우기보다는, 내 서비스에 필요한 일 중 각 도구가 어디까지 맡아주는지 살펴보세요.

### 도구마다 해결하려는 문제가 다릅니다

프레임워크는 반복 작업을 줄이는 방법과 제공 범위가 서로 다릅니다. 화면을 만드는 프런트엔드 도구를 비교할 때는 다음 세 가지를 살펴볼 수 있습니다.

| 관리할 일 | 직접 정해야 할 내용 | 도구가 도울 수 있는 부분 |
|---|---|---|
| 반복 화면 | 공통 코드를 어디에 두고 어떻게 재사용할지 | 컴포넌트 구성 규칙 |
| 데이터 변화 | 바뀐 값이 영향을 주는 화면을 어떻게 갱신할지 | 상태와 화면을 연결하는 규칙 |
| 여러 화면과 협업 | 주소, 파일, 테스트를 어떻게 관리할지 | 라우팅과 프로젝트 구성 도구 |

표에 나온 세 개념을 화면의 예와 연결해 봅시다.

- **컴포넌트** 는 카드, 검색창처럼 따로 구성하고 재사용할 수 있는 화면 단위입니다.
- **상태(state)** 는 검색어, 선택한 분류, 로딩 여부처럼 화면에 영향을 주는 현재 값입니다.
- **라우팅** 은 주소에 맞는 화면을 연결하는 기능입니다.

모든 도구가 이 기능들을 기본으로 제공하는 것은 아닙니다. 필요한 기능을 어디까지 제공하는지 확인해야 합니다.

### 상태를 기준으로 화면 표현하기

검색어가 바뀌면 카드 목록과 결과 수가 함께 바뀌어야 합니다. 이때 검색어를 상태로 두고, 그 값에 따라 어떤 카드와 결과 수를 보여줄지 작성할 수 있습니다. 도구는 그 결과에 맞춰 화면 갱신을 처리합니다. 이런 방식을 **선언형 UI** 라고 합니다.

도구가 갱신을 맡아주더라도, 어떤 값을 상태로 관리하고 어느 컴포넌트가 그 값을 사용할지는 정해야 합니다. 다음 사례에서 직접 관리할 일과 도구가 도와줄 일을 구분해 봅시다.

### 반복 카드는 순수 JavaScript로도 재사용할 수 있습니다

독서 목록에 책 카드가 30개 있다고 해 봅시다. 카드마다 HTML을 복사해 두었다면 표시 형식을 바꿀 때 여러 곳을 수정해야 합니다. 하지만 순수 JavaScript도 함수를 만들어 카드 구조를 한 곳에서 관리하고, 배열의 데이터로 반복 생성할 수 있습니다.

React의 컴포넌트도 카드 구조를 한 곳에서 관리합니다. 따라서 카드를 재사용한다는 사실만으로 React가 필요한 것은 아닙니다.

차이는 검색어가 바뀌는 상황에서 살펴볼 수 있습니다. 순수 JavaScript로 구성한다면 목록을 바꾸는 코드와 결과 수를 바꾸는 코드가 함께 실행되도록 직접 연결해야 합니다. React에서는 같은 검색어 상태를 기준으로 목록과 결과 수를 표현하고, 화면에 변경을 적용하는 처리는 React에 맡깁니다.

| 작업 | 순수 JavaScript로 직접 구성 | React 사용 |
|---|---|---|
| 카드 재사용 | 카드 생성 함수와 데이터 배열 사용 | 카드 컴포넌트와 데이터 배열 사용 |
| 검색어 변경 | 관련 목록과 결과 수의 갱신 코드를 연결 | 상태를 기준으로 목록과 결과 수를 표현 |
| 팀 작업 | 파일 구조와 갱신 규칙 등을 합의 | 도구의 규칙을 참고하고 프로젝트 규칙을 추가 |

프레임워크를 사용해도 상태 설계와 팀의 합의는 필요합니다. 단순한 화면이라면 직접 구성하는 편이 관리하기 쉬울 수도 있습니다.

### 상태가 바뀌면 어디까지 함께 바뀌어야 할까요?

동아리 일정 목록에서 사용자가 ‘이번 주’를 선택하면 목록과 일정 수가 함께 바뀌어야 합니다. 목록만 바뀌고 수가 그대로라면 화면의 정보가 서로 맞지 않습니다.

AI에게는 다음처럼 관찰할 수 있는 동작을 요청할 수 있습니다.

> 선택한 기간을 하나의 상태로 관리하고, 그 값으로 일정 목록과 일정 수를 계산해 주세요. 기간을 바꿀 때 두 표시가 함께 바뀌는지 확인해 주세요.

도구를 선택할 때도 이런 동작이 얼마나 많은지, 함께 수정할 사람이 어떤 규칙을 공유할 수 있는지 살펴봅니다.

### 일정 화면으로 이해하는 추상화와 모듈화

방금 일정 화면에서는 선택한 기간에 따라 무엇을 보여줄지 정하고, 실제 화면을 바꾸는 세부 처리는 도구에 맡겼습니다. 이렇게 사용할 때 필요한 동작을 드러내고 내부 처리의 세부 사항은 감추는 것을 **추상화** 라고 합니다.

같은 화면을 검색창과 일정 카드로 나누고, 각각 어떤 데이터를 받아 무엇을 보여줄지 정할 수도 있습니다. 이처럼 프로그램을 역할과 경계가 있는 단위로 나누는 것이 **모듈화** 입니다. 경계를 잘 정하면 한 부분을 수정할 때 다른 부분에 미치는 영향을 줄이고, 개별 기능을 확인하기 쉽습니다.

<details class="lms-extra"><summary>파나스의 정보 은닉 원칙 (선택)</summary>

일정의 저장 방식을 바꿔야 한다고 해 봅시다. 화면 코드 곳곳에 저장 방법이 섞여 있다면 그 코드도 함께 고쳐야 합니다. 반대로 저장을 담당하는 부분을 따로 두고 화면에서는 필요한 동작만 요청하게 만들면, 저장 부분을 고치면서 화면 코드는 그대로 둘 수 있습니다.

이처럼 바뀔 수 있는 처리 방법을 한 부분 안에 두고, 다른 코드가 그 세부 사항을 몰라도 동작하도록 나누는 기준을 **정보 은닉** 이라고 합니다. David Parnas는 1972년 논문 「On the Criteria To Be Used in Decomposing Systems into Modules」에서 이를 모듈 분리 기준으로 제시했습니다.

### React가 화면 변경을 적용하는 과정

일정 화면에서 도구에 맡긴 갱신 처리는 어떻게 이루어질까요? React를 예로 살펴보겠습니다.

**DOM(Document Object Model)** 은 브라우저가 문서의 요소와 구조를 객체로 표현한 것입니다. 일정 목록과 일정 수를 표시하는 요소도 이 구조에 포함됩니다. JavaScript는 DOM을 통해 화면 내용을 바꿀 수 있습니다.

React는 실제 DOM을 바로 바꾸기 전에, 화면이 어떤 모습이어야 하는지 메모리 안의 표현으로 계산합니다. 이를 설명할 때 **가상 DOM** 이라는 용어를 사용합니다. 사용자가 기간을 ‘이번 주’로 바꿨을 때의 과정을 따라가 봅시다.

1. **계산**: 바뀐 기간 상태를 바탕으로 컴포넌트의 코드를 실행해, 보여줄 일정 목록과 일정 수가 담긴 새 화면 결과를 계산합니다. React에서는 이 계산을 **렌더링** 이라고 합니다.
2. **비교**: 이전에 계산한 화면 결과와 새 결과를 비교해 목록과 일정 수에서 무엇이 달라졌는지 확인합니다. 매번 실제 DOM 전체를 읽어 비교한다는 뜻은 아닙니다.
3. **적용**: 필요한 변경을 실제 DOM에 반영합니다. React 공식 문서는 이 단계를 **커밋** 이라고 구분합니다. 계산 결과가 이전과 같다면 DOM을 바꾸지 않을 수도 있습니다.

이 과정은 React의 화면 갱신을 이해하기 위한 예이지, 모든 도구가 같은 방식을 쓴다는 뜻은 아닙니다. 가상 DOM의 사용만으로 속도 우위를 판단할 수도 없습니다. 성능은 화면의 구성과 갱신 방식에 따라 달라지므로 실제 서비스 조건에서 측정해야 합니다.

- [Svelte 공식 소개](https://svelte.dev/)

### 요구사항으로 판단해 보기

이제 도구의 내부 처리보다, 내 화면에서 관리해야 할 일에 초점을 맞춰 봅시다. 앞에서 책 카드는 한 곳에서 구조를 관리할 수 있었고, 일정 화면은 선택한 기간에 따라 목록과 일정 수가 함께 바뀌어야 했습니다.

이 두 사례를 떠올리며 다음 요구사항에서 프레임워크가 줄일 수 있는 작업과, 오히려 설정·관리가 부담이 될 상황을 적어 보세요.

- 필터, 목록, 결과 수처럼 여러 표시가 같은 상태에 의존하면 갱신 규칙의 도움을 받을 수 있습니다.
- 정적 안내 페이지는 프레임워크의 설정과 관리 비용이 이점보다 클 수 있습니다.
- 함께 오래 유지할 화면은 문서, 업데이트 정책, 팀의 경험과 작성 규칙도 비교해야 합니다.

</details>

## 내 서비스에 맞는 프레임워크 선택 기준

프레임워크를 고를 때는 만들 화면과 운영 조건을 먼저 적습니다. 필요한 기능, 팀의 경험, 배포 환경을 정리하면 후보를 비교할 기준이 생깁니다. 이름뿐 아니라 함께 쓸 라이브러리, 문서, 업데이트 정책도 살펴봅니다.

후보를 비교하기 전에 도구가 맡는 역할을 구분해 보겠습니다. **메타프레임워크** 는 UI 라이브러리·프레임워크를 바탕으로 라우팅, 서버 렌더링 등 애플리케이션 구성에 필요한 기능을 묶은 도구입니다. 서버 렌더링은 서버에서 화면에 보여 줄 HTML을 만드는 방식이며, 서버 컴포넌트는 서버에서 실행하는 화면 단위, 클라이언트 컴포넌트는 브라우저에서 클릭 같은 상호작용을 처리할 수 있는 화면 단위입니다.

**빌드** 는 앞서 본 컴파일처럼 코드를 변환하는 작업을 포함해 배포할 파일을 만드는 과정입니다.

| 범주 | 예시 | 맡는 역할 |
|---|---|---|
| UI 라이브러리·프레임워크 | React, Vue, Svelte, Angular | 화면을 컴포넌트로 구성하고 데이터 변화에 따라 갱신합니다 |
| 메타프레임워크 | Next.js, Nuxt, SvelteKit | 기반 UI 도구에 라우팅·서버 렌더링 등을 더합니다 |
| 개발·빌드 도구 | Vite | 개발 서버를 제공하고 배포할 파일을 만듭니다 |
| 함께 쓰는 도구 | 라우터, 상태 관리, UI 컴포넌트, 테스트 도구 | 프로젝트에 필요한 기능을 보완합니다 |

[React 공식 문서](https://react.dev/)는 React를 UI 라이브러리로 소개하고, 컴포넌트를 조합해 화면을 만드는 방법을 설명합니다. [Next.js의 App Router 문서](https://nextjs.org/docs/app)는 화면 파일을 둔 위치에 따라 주소를 연결하는 라우팅과 서버·클라이언트 컴포넌트를 설명합니다. React를 선택한 뒤 Next.js를 함께 사용할지 검토할 수 있습니다.

[Vite 공식 문서](https://vite.dev/guide/)는 개발 서버와 빌드를 제공하는 도구라고 설명합니다. 따라서 React와 Vite를 함께 쓸 수도 있습니다. AI가 도구를 추천하면 각 도구의 역할과 함께 필요한 구성을 설명해 달라고 요청하세요. 같은 역할을 맡는 후보끼리 비교해야 선택 이유를 분명히 적을 수 있습니다.

### 조사 결과를 선택 근거로 읽는 방법

도구의 역할을 구분했다면, 조사 결과가 후보를 고르는 데 어떤 도움을 주는지 살펴보겠습니다. 사용 경험을 묻는 조사와 AI가 만든 코드를 평가한 연구는 서로 다른 정보를 보여 줍니다.

#### 사용률에서 알 수 있는 것

프레임워크의 사용률은 후보를 찾는 참고 자료입니다. Stack Overflow Developer Survey 2025의 웹 프레임워크·기술 문항은 지난 1년간 개발에 사용한 기술과 다음 1년에 사용하고 싶은 기술을 따로 묻습니다. 결과를 읽을 때는 사용 경험인지 사용 희망인지, 전체 응답자인지 전문 개발자인지 확인해야 합니다. 사용률만으로 내 서비스에 필요한 기능이나 유지보수 비용까지 알 수는 없습니다. ([조사 원문](https://survey.stackoverflow.co/2025/technology/))

Stack Overflow 2025 조사에서 해당 문항의 전체 응답자(All Respondents) 기준 사용률은 React 44.7%, Angular 18.2%, Vue.js 17.6%, Next.js 20.8%입니다. 모두 지난 1년간의 개발 경험을 묻는 같은 집단의 수치입니다. 같은 표에 jQuery 23.4% 등도 포함되므로 이 네 기술만 보고 전체 순위를 매기지는 않습니다. ([조사 원문](https://survey.stackoverflow.co/2025/technology/))

이 수치는 AI 학습 데이터의 양이나 생성 코드의 정확성을 직접 측정한 값이 아닙니다. React와 Next.js는 함께 사용할 수 있으므로 두 수치를 서로 배타적인 선택 비율로 읽어서도 안 됩니다. 사용률은 후보를 찾는 참고 자료로 두고, 내 요구사항에 맞는지는 따로 확인합니다.

#### AI가 만든 코드에서 확인할 것

AI는 요청한 프레임워크로 코드 초안을 만들 수 있지만, 문법과 화면 동작, 관리하기 좋은 구조까지 보장하지는 않습니다. 연구 결과를 읽을 때도 무엇을 성공으로 판단했는지 먼저 봐야 합니다.

**컴파일 성공** 은 코드를 실행 가능한 형태로 변환하는 데 성공했다는 뜻입니다. 원하는 화면과 기능을 완성했다는 뜻까지 포함하지는 않습니다.

프런트엔드 코드 생성을 평가한 DesignBench 논문의 **디자인 생성 과제** 에서 상위 모델들의 컴파일 성공률은 React·Vue에서 0.83을 넘었고, Angular에서는 약 0.69～0.76이었습니다. 여기서 성공률은 전체 평가 중 컴파일에 성공한 비율을 소수로 나타낸 값입니다. 이는 해당 과제와 평가 모델에서 얻은 결과입니다. 모든 AI 모델이나 모든 웹서비스 개발 상황에 그대로 적용할 수는 없습니다. ([논문 VI-A](https://arxiv.org/html/2506.06251v1))

같은 논문은 프레임워크별 문법 오류와 재사용 가능한 컴포넌트를 충분히 활용하지 못하는 한계도 보고합니다. 이 결과만으로 학습 데이터의 양이 성능 차이의 원인이라고 결론 내릴 수는 없습니다.

따라서 AI가 작성한 상품 목록 화면도 다음 세 가지를 나누어 확인합니다.

- 컴파일을 포함한 빌드 과정이 성공해 배포할 파일이 만들어지는가?
- 요청한 상품 정보가 화면에 표시되는가?
- 반복되는 상품 표시를 재사용 가능한 컴포넌트로 구현했는가?

#### 내 조건으로 후보를 비교하는 방법

후보를 2～3개로 좁힌 뒤, 같은 화면 요구사항과 같은 AI 모델로 작은 구현을 요청해 볼 수 있습니다. 도구 버전, 입력 데이터, 표시할 정보와 동작 조건을 동일하게 적고 실행 결과와 수정한 내용을 기록합니다. 이 비교는 내 작업에서 필요한 검토량을 파악하기 위한 참고 자료입니다. 몇 번의 생성 결과만으로 특정 프레임워크의 정확성을 일반화하지는 않습니다.

공식 문서에서 제시하는 구현 방법과 비교하고, 오류를 고치는 데 필요한 설명과 수정량도 살펴보세요. 프레임워크의 사용률, AI의 생성 결과, 팀이 이해하고 관리할 수 있는 범위는 각각 따로 확인할 정보입니다.

### 후보를 비교할 여섯 가지 기준

AI 에이전트로 서비스를 만들고 관리하려면 필요한 조건을 설명하고 추천의 근거를 검토해야 합니다. 다음 기준으로 후보의 장점과 부담을 함께 적어 보세요.

| 선택 기준 | 확인할 질문 | 살펴볼 이유 |
|---|---|---|
| AI 생성 결과의 검토 부담 | 같은 요구사항으로 만든 코드가 실행되고, 필요한 수정을 설명할 수 있나요? | 초안의 오류와 수정량을 확인하면 작업에 필요한 검토 시간을 예상할 수 있습니다 |
| 학습 부담 | 팀이 필요한 개념을 익히고 코드를 읽으며 수정할 수 있나요? | 시작할 때 드는 시간과 이후 관리에 필요한 이해를 함께 고려합니다 |
| 생태계 | 필요한 인증, 차트, 폼, UI, 배포 예제를 구할 수 있나요? | 직접 구현할 범위와 기존 도구를 활용할 범위를 정하는 데 도움이 됩니다 |
| 인력·자료 | 유지보수를 맡을 사람과 참고할 문서·사례를 확보할 수 있나요? | 담당자가 바뀌어도 개발과 운영을 이어 갈 수 있는지 확인합니다 |
| 장기 유지보수 | 업데이트 정책과 보안 패치 주기를 서비스 운영 조건에 맞출 수 있나요? | 도입 이후에도 업데이트와 변경 대응이 필요합니다 |
| 트레이드오프 | 얻는 이점과 함께 늘어나는 복잡도·배포 비용을 감당할 수 있나요? | 한 기능을 얻기 위해 추가되는 작업과 비용을 함께 비교합니다 |

모든 기준에 같은 비중을 줄 필요는 없습니다. 서비스의 일정과 운영 조건에 맞춰 우선순위를 정하고, 후보 2～3개의 채택·기각 이유를 남깁니다.

### Next.js를 후보에 넣을 때 확인할 점

앞에서 살펴본 Next.js를 후보에 넣었다면, 제공하는 기능을 내 서비스에서 얼마나 쓸지 따져 봅니다. 공식 문서에는 데이터 가져오기, 오류 처리, 배포 등도 안내되어 있습니다. ([공식 문서](https://nextjs.org/docs/app))

여러 페이지와 서버에서 처리할 기능이 필요하다면 이 기능들이 서비스 요구에 맞는지 확인합니다. 단순한 소개 페이지라면 추가로 익힐 개념과 배포 구성이 필요한지도 비교하세요. 로그인이나 데이터 처리는 선택만으로 완성되지 않으므로, 필요한 인증 도구와 데이터 저장 방식까지 별도로 검토합니다.

## 내 서비스의 선택 근거와 PRD 만들기

### AI와 대화하고 결과 저장하기

#### 1. 내 서비스 인터뷰하기

브라우저에서 ChatGPT나 Claude의 새 대화를 엽니다. 아래 대괄호를 내 내용으로 바꿔 입력하고, AI의 질문에 답합니다.

```text
[만들고 싶은 서비스]로 [사용자]의 [해결할 문제]를 해결하고 싶어.
서비스 목적, 핵심 화면, 필요한 기능과 내 개발 경험·시간을
한 번에 하나씩 물어봐 줘. 아직 정하지 않은 내용은 따로 표시해 줘.
```

#### 2. 후보 비교하고 선택 근거 쓰기

같은 대화에 아래 요청을 붙여 넣습니다. 앞 활동표에서 고른 우선 기준 2개도 넣습니다.

```text
내 조건에 맞는 프레임워크 후보 2～3개를 추천해 줘.
AI 생성 결과의 검토 부담, 학습 부담, 생태계, 인력·자료,
장기 유지보수, 트레이드오프의 여섯 기준으로 비교해 줘.
내 우선 기준은 [기준 1]과 [기준 2]야.
확인이 필요한 정보는 표시하고, 각 후보의 채택·기각 이유를 제안해 줘.
```

비교를 읽고 내 조건과 다른 부분을 고칩니다. 채택할 후보 하나와 기각할 후보를 정해 AI에게 알려주고, **선택 근거** 를 다음 항목으로 정리해 달라고 요청합니다.

```text
내 선택은 [후보]이고 이유는 [내 조건과 연결한 이유]야.
다른 후보를 기각하는 이유는 [후보별 이유]야.
선택 근거를 '내 조건', '후보 2～3개 비교', '채택·기각 이유'로 정리해 줘.
```

#### 3. PRD 작성하고 파일로 남기기

이어서 아래 요청을 입력합니다. 아직 화면이나 목업 데이터는 만들지 않습니다.

```text
지금까지 합의한 내용으로 PRD를 작성해 줘. 다음 항목을 모두 넣어 줘.
- 서비스 목적: 누구의 어떤 문제를 해결하는가
- 핵심 화면: 화면별 역할과 이동 흐름
- 요구사항: 사용자가 할 수 있어야 하는 일과 확인할 조건
- 이번 구현 범위: 목업 데이터로 구현할 화면·동작과,
  나중에 백엔드가 필요한 기능을 구분
선택한 프레임워크와 선택 근거도 반영해 줘.
```

컴퓨터에 이 서비스의 작업 폴더를 만듭니다. 텍스트 편집기에서 선택 근거와 PRD를 각각 붙여 넣고, 같은 폴더에 `선택-근거.md`와 `PRD.md`로 저장합니다. 두 파일은 과제에서도 사용합니다.

Mac의 텍스트편집에서는 ‘서식 > 일반 텍스트 만들기’로 바꾼 뒤 저장할 때 파일 이름에 `.md` 확장자를 입력하고, `.txt`가 덧붙지 않았는지 확인하세요.

저장한 파일을 열어 다음을 확인하고, 빠진 내용은 보완합니다.

- [ ] 내 조건과 우선 기준 2개가 선택 근거에 있습니다.
- [ ] 후보 2～3개를 여섯 기준으로 비교했고, 각각 채택·기각 이유가 있습니다.
- [ ] PRD에 서비스 목적, 핵심 화면, 요구사항, 이번 구현 범위가 있습니다.
- [ ] 이번 구현 범위에서 목업 데이터로 확인할 것과 백엔드가 필요한 것을 구분했습니다.

## 백엔드 없이 프런트엔드 만들기

### 목업 데이터로 화면 확인하기

여기서부터는 내 서비스 대신 수제 디저트 쇼핑몰을 예시로 함께 따라 해 봅니다. 예시 실습을 마친 뒤 과제에서 다시 내 서비스로 돌아와 목업 데이터가 필요한지 판단합니다.

앞 레슨에서 본 프런트엔드와 백엔드를 식당에 비유하면, 프런트엔드는 손님이 보는 홀이고 백엔드는 주문을 처리하는 주방입니다. 화면 표시와 데이터 처리는 서로 다른 역할입니다.

**목업 데이터** 는 실제 데이터 연결 전에 화면을 확인하는 데 쓰는 예시 데이터입니다. 쇼핑몰에서는 상품명과 가격 같은 정보를 미리 준비해 목록과 상세 화면, 버튼을 눌렀을 때의 변화를 확인할 수 있습니다.

실제 상품 데이터를 연결할 때는 **API** 를 사용합니다. API는 프로그램이 다른 프로그램에 기능이나 데이터를 요청할 때 따르는 규칙입니다. 예를 들어 프런트엔드가 상품 목록을 요청하면 백엔드는 정해진 형식으로 응답합니다. 이번에는 그 응답 대신 목업 데이터를 준비합니다.

웹에서 요청과 응답을 주고받을 때 사용하는 통신 규칙을 **HTTP** 라고 합니다. 프런트엔드는 이 규칙에 따라 백엔드에 상품 목록을 요청하고 응답을 받습니다. 응답이 JSON 형식이면 이를 JavaScript에서 사용할 데이터로 변환해 화면에 표시합니다.

### 상품 데이터가 화면에 표시되는 과정

웹 API에서는 **JSON** 형식을 자주 씁니다. 아래는 상품 객체 두 개를 담은 JSON 배열입니다. 이 레슨 뒤의 수제 디저트 쇼핑몰 실습도 같은 구조를 사용합니다.

```json
[
  { "id": 1, "name": "초콜릿 쿠키", "price": 3500, "category": "쿠키", "image": "/images/chocolate-cookie.jpg", "soldOut": false },
  { "id": 2, "name": "딸기 타르트", "price": 6500, "category": "타르트", "image": "/images/strawberry-tart.jpg", "soldOut": true }
]
```

화면은 이 목록에서 상품 정보를 하나씩 읽어 같은 모양의 카드로 표시합니다. 첫 번째 정보로는 ‘초콜릿 쿠키’ 카드를, 두 번째 정보로는 ‘딸기 타르트’ 카드를 만듭니다. `price`는 원 단위 숫자이므로 각 카드의 가격은 3500원과 6500원입니다.

`soldOut`은 품절 여부를 나타냅니다. `false`인 초콜릿 쿠키는 품절이 아니고, `true`인 딸기 타르트에는 품절 표시가 필요합니다. 이처럼 참(`true`)과 거짓(`false`) 두 값으로 나타내는 데이터를 **불리언** 이라고 합니다. `image`는 이미지 경로이며, 해당 파일을 준비해야 사진이 표시됩니다.

이 예시의 상품명, 가격, 품절 여부처럼 화면에서 사용할 데이터 항목을 먼저 정해 두세요. 목업 데이터와 API 응답을 이 구조로 맞추면 데이터를 가져오는 부분만 바꾸고 화면 컴포넌트는 거의 그대로 둡니다. API를 연결할 때는 로딩·오류 처리도 함께 추가합니다.

### 확인할 조건을 목업 데이터에 담기

상품 카드가 표시되는 것만으로 확인을 끝내지는 않습니다. PRD에 적은 화면 조건에 맞춰 데이터를 바꾸고, 다음 상황에서도 화면이 의도대로 동작하는지 살펴봅니다.

- 긴 상품명을 넣어 글자가 카드 밖으로 나가는지 확인합니다.
- `soldOut`이 `true`인 상품을 넣어 품절 표시와 버튼 동작을 확인합니다.
- 상품 배열을 `[]`로 바꿔 목록이 비었을 때 안내가 나오는지 확인합니다.

목업 데이터는 `data/products.json`에 JSON 배열로 저장합니다. 상품명이나 가격 같은 데이터 항목을 **필드** 라고 합니다. 각 상품에 쓸 필드 이름은 앞의 예시처럼 `id`, `name`, `price`, `category`, `image`, `soldOut`으로 통일합니다. 데이터를 여러 화면 안에 따로 적지 않고 한 파일에서 가져오면, 값을 수정해 같은 조건으로 다시 확인하기 쉽습니다.

이 단계에서 확인하는 것은 화면 표시와 상호작용입니다. 서버에 저장되는지, 요청 권한이 맞는지, 연결 실패를 처리하는지는 API를 연결한 뒤 별도로 확인합니다.

### AI에게 목업 데이터 요청하기

PRD에서 상품 목록에 필요한 조건을 골라 AI에 전달합니다. 아래는 화면에 표시할 필드와 확인할 상황을 함께 적은 요청 예시입니다. 이어지는 수제 디저트 실습에서는 이 방식을 써서 파일을 직접 만듭니다.

**JSON** 은 이름과 값의 짝으로 데이터를 적는 형식이며, 상품 하나를 `{}`로 묶고 여러 상품을 `[]` 안에 나열하면 JSON 배열이 됩니다. 품절 여부에 쓰는 **불리언** 은 `true` 또는 `false`로 나타내는 값입니다. 여기서는 `"soldOut": true`가 품절, `"soldOut": false`가 판매 중을 뜻합니다.

```text
PRD의 상품 목록 조건에 맞춰 수제 디저트 쇼핑몰의 목업 데이터를 만들어 줘.
data/products.json에 저장할 JSON 배열로 작성해 줘.
각 객체에는 id, name, price, category, image, soldOut을 넣어 줘.
id는 중복 없는 숫자, name과 category는 문자열로 작성해 줘.
price는 원 단위 숫자, soldOut은 불리언으로 작성해 줘.
image는 프로젝트에서 사용할 이미지 경로 문자열로 넣어 줘.
긴 상품명과 품절 상품을 포함하고 JSON 안에 주석은 넣지 마.
빈 목록 확인용 배열은 본문 배열과 별도로 제시해 줘.
```

응답을 받으면 바깥이 배열인지, 상품마다 여섯 필드가 있는지 확인합니다. 가격에 `"3,500원"`처럼 문자가 들어가거나 품절 여부가 `"false"`처럼 문자열로 나오면 수정을 요청합니다. 이미지 경로와 실제 준비한 파일도 맞춥니다.

빈 목록을 확인할 때는 상품 하나의 내용을 비우는 대신 배열 전체를 `[]`로 바꿉니다.

연결 후에는 정상 목록, 빈 목록, 느린 응답, 실패 응답을 각각 확인합니다. 목업 화면이 동작했다는 사실만으로 API 연결까지 확인한 것은 아닙니다.

## CS 교양: API 요청과 응답 이해하기

목업 화면을 완성한 뒤 실제 서버와 연결할 때 참고할 내용입니다. 지금 화면을 만드는 중이라면 다음 「수제 디저트 상품 목록 화면 만들기」로 넘어가세요.

상품 목록을 서버에서 받으려면 먼저 `GET /api/products`처럼 주소와 요청 방법을 보냅니다. 서버는 처리 결과를 나타내는 상태 코드와 상품 데이터를 응답하고, 화면은 그 데이터를 읽어 목록을 표시합니다. **API** 는 이때 사용할 주소, 요청 방법, 입력값과 응답 형식을 정합니다.

이번 예시에서는 응답 본문으로 상품들의 JSON 배열을 받기로 합니다. 모양만 간단히 비교하면 `[{"id": 1, "name": "초콜릿 쿠키"}]`는 배열 자체이고, `{"products": [{"id": 1, "name": "초콜릿 쿠키"}]}`는 배열을 `products`라는 이름으로 한 번 더 묶은 형태입니다. 뒤의 형태라면 응답 전체가 아니라 `products` 안의 배열을 꺼내 화면에 전달해야 합니다. 실제 상품에는 목업과 같은 여섯 필드가 모두 있어야 합니다.

연결할 때는 아래의 요청·응답 설명부터 확인하세요. API 방식 비교와 데이터 형식의 배경은 필요할 때 펼쳐 읽으면 됩니다.

<details class="lms-extra"><summary>HTTP 요청과 상태 코드 자세히 보기 (선택)</summary>

#### 요청에 담는 정보

요청에는 주소인 URL, 작업을 나타내는 메서드, 데이터 형식 같은 부가 정보를 담는 헤더가 있고 필요에 따라 본문을 보냅니다. URL의 경로는 대상을 가리키고, 물음표 뒤의 쿼리 매개변수는 검색 조건 등을 전달합니다. `GET /api/products?category=쿠키`라면 상품 중 쿠키에 해당하는 목록을 요청한다는 뜻으로 정할 수 있습니다.

주소 구조를 보여 주는 기존 예시 `https://내서비스.com/api/experts?specialty=형사`도 같은 방식입니다. 경로는 `/api/experts`, 검색 조건은 `specialty=형사`입니다. 설명용 주소이며 상품 실습에서 호출할 API는 아닙니다.

| 메서드 | 주로 나타내는 작업 | 상품 API의 예 |
|---|---|---|
| GET | 데이터 조회 | 상품 목록 조회 |
| POST | 데이터 생성 또는 처리 요청 | 새 상품 등록 |
| PUT | 대상의 표현 전체 교체 | 상품 정보 전체 교체 |
| PATCH | 대상의 일부 변경 | 상품 가격 변경 |
| DELETE | 대상 삭제 | 상품 삭제 |

이 표는 일반적인 사용 방식입니다. 실제 주소와 허용하는 작업, 필수 입력값은 서버의 API 명세에서 확인합니다. 화면에 삭제 버튼이 있다고 해서 서버가 삭제 요청을 허용하는 것은 아닙니다.

#### 응답에서 확인할 정보

응답의 상태 코드는 처리 결과를 나타냅니다. `200`은 성공, `201`은 생성 성공에 사용합니다. `400`은 잘못된 요청, `404`는 대상을 찾지 못한 경우, `500`은 서버 내부 오류를 나타냅니다. 응답 본문에는 상품 배열이나 오류 내용 등이 들어갈 수 있습니다. 모든 응답에 JSON 본문이 있는 것은 아닙니다.

서버에 연결하지 못한 경우와, 연결은 됐지만 서버가 `404`나 `500`을 돌려준 경우는 다릅니다. 브라우저에서 요청을 보내는 `fetch`는 뒤의 경우를 자동으로 실패 처리하지 않으므로, 응답의 `ok` 값이나 상태 코드를 확인해야 합니다. AI가 만든 코드에서도 두 경우 모두 오류 안내를 표시하는지 확인하세요. 성공한 응답만 상품 배열로 읽어야 오류 내용을 상품 목록으로 잘못 처리하지 않습니다.

</details>

<details class="lms-extra"><summary>REST·GraphQL·RPC·SOAP 비교하기 (선택)</summary>

상품 목록을 요청하는 방법에는 어떤 차이가 있을까요? 프로토타입에서는 조회할 대상을 URL로 나타내는 REST 방식의 HTTP API부터 검토하되, 기존 서버가 있다면 그 규칙을 먼저 따릅니다. REST를 선택해도 JSON 배열 같은 세부 응답 형식은 별도로 정해야 합니다.

| 방식 | 요청을 구성하는 방법 | 선택할 때 확인할 점 |
|---|---|---|
| REST | URL로 대상을 나타내고 HTTP 메서드 등 웹의 규칙을 활용 | 자원별 조회·변경을 표현하기 좋은지 |
| GraphQL | 클라이언트가 필요한 필드를 질의로 지정 | 화면마다 필요한 데이터가 다른지, 서버가 질의를 지원하는지 |
| RPC | 원격의 기능을 호출할 이름과 입력값을 전달 | 데이터 목록보다 특정 작업 호출이 중심인지 |
| SOAP | XML 메시지와 정해진 메시지 처리 규칙을 사용 | 연동 대상이 SOAP 규약을 요구하는지 |

REST는 특정 데이터 형식의 이름이 아니라 설계 방식입니다. GraphQL은 질의 언어와 실행 규칙을 제공하며, RPC는 원격 호출 방식의 넓은 범주입니다. SOAP는 메시지 교환 프로토콜입니다. 서로 완전히 같은 종류의 기술을 나열한 표는 아닙니다.

GraphQL은 필요한 필드를 선택할 수 있지만 서버의 질의 처리와 권한 검사도 설계해야 합니다. RPC는 특정 작업을 직접 표현하기 좋고, SOAP는 기존 연동 시스템의 요구로 선택할 수 있습니다. 어느 방식이 항상 더 빠르거나 더 간단하다고 단정하지 않고, 화면에 필요한 요청과 연결할 서버를 기준으로 결정합니다.

</details>

<details class="lms-extra"><summary>API와 데이터 형식의 변화 살펴보기 (선택)</summary>

여러 API 방식이 함께 쓰이는 이유는 무엇일까요? 위 방식들은 서로 다른 요구에서 발전했으며, 새로운 방식이 등장해도 기존 시스템과의 연결 조건에 따라 이전 방식을 계속 사용합니다.

XML은 태그로 요소와 문서 구조를 표현합니다. JSON은 객체·배열·문자열·숫자·불리언·null로 데이터를 표현합니다. 같은 상품명도 XML에서는 `<name>초콜릿 쿠키</name>`처럼, JSON 객체에서는 `{"name": "초콜릿 쿠키"}`처럼 적을 수 있습니다.

JSON은 JavaScript 객체 표기에서 유래했지만 JavaScript 실행 코드가 아닙니다. 브라우저에서도 받은 텍스트를 JSON으로 해석하는 과정이 필요합니다. 여러 프로그래밍 언어가 JSON을 지원하므로 프런트엔드와 백엔드가 서로 다른 언어를 써도 데이터를 주고받을 수 있습니다.

JSON이 모든 용도에서 XML을 대체한 것은 아닙니다. 문서 구조와 기존 시스템의 규약에 따라 XML을 쓰기도 합니다. 화면 개발에서는 형식의 우열보다 어떤 응답이 오는지, 필요한 필드와 자료형이 맞는지 확인하는 일이 먼저입니다.

</details>

## 수제 디저트 상품 목록 화면 만들기

수제 디저트 쇼핑몰의 조건과 선택 근거를 예시로 읽고, 목업 데이터로 상품 목록 화면을 만들어 봅니다. 앞에서 자신의 서비스로 만든 선택 근거와 PRD는 과제에서 이어 씁니다.

### 예시의 조건과 작업 범위

| 항목 | 수제 디저트 쇼핑몰의 조건 |
|---|---|
| 이용자와 목적 | 선물할 디저트를 찾는 사람이 상품을 살펴봅니다 |
| 이번 화면 | 상품 6개의 이름·가격·분류·이미지 영역과 품절 여부를 보여 줍니다 |
| 선택 조건 | 공개 상품 페이지의 검색 노출을 고려하고, React 자료·예제를 참고하며 개발합니다 |
| 향후 기능 | 주문 내역 저장과 재고 관리는 백엔드와 데이터베이스가 필요합니다 |
| 지금 만들 범위 | 로컬 목업 파일로 상품 목록만 만듭니다. 백엔드·주문·결제는 구현하지 않습니다 |

검색 노출은 프레임워크 선택 때 고려할 조건입니다. 이번 로컬 실습에서는 실제 검색 노출까지 검증하지 않습니다.

### 예시의 선택 근거

| 후보 | 조건에 맞는 점 | 이번 선택에서 고려한 제약 |
|---|---|---|
| Next.js | React를 사용하며, 공개 상품 페이지를 서버에서 렌더링하거나 미리 생성할 수 있습니다 | 목록만 만드는 데에는 설정과 개념이 더 필요합니다 |
| React + Vite | React 자료를 참고하면서 브라우저용 목록 화면을 만들 수 있습니다 | 검색 노출을 위한 렌더링 방식은 추가로 검토해야 합니다 |

이 예시에서는 Next.js를 선택합니다. React를 사용한다는 조건에 더해, 검색 서비스가 읽을 공개 상품 정보를 HTML로 준비하는 방식을 고려했습니다. Next.js는 서버에서 상품 내용이 담긴 HTML을 만들거나 미리 생성할 수 있어, 공개 상품 페이지로 확장할 때 활용할 수 있습니다. 이것만으로 검색 노출을 보장하는 것은 아니며, 이번 로컬 실습의 검증 대상도 아닙니다. React + Vite도 목업 목록에는 맞지만, 공개 상품 페이지까지 같은 도구로 확장할 계획을 고려해 이번에는 채택하지 않았습니다.

선택한 도구로 무엇을 만들지도 좁힙니다. 예시 PRD의 구현 범위는 “목업 파일의 상품 6개를 목록으로 표시하고 품절 상품을 구분한다”입니다. Next.js의 서버 기능을 사용할 수 있어도 이번 실습에는 백엔드를 만들지 않습니다.

### 1단계: 실습 폴더에서 목업 데이터 만들기

자신의 컴퓨터에서 앞 레슨처럼 문서 폴더 등 프로젝트를 모아 두는 위치에 새 `dessert-shop` 폴더를 만드세요. 같은 이름의 폴더가 있으면 `dessert-shop-practice`처럼 겹치지 않는 이름을 사용합니다. 앞 레슨의 `my-first-web`·`my-game` 폴더는 그대로 둡니다.

Claude Desktop에서는 Code의 Local 작업 폴더로 새 폴더를 선택합니다. 터미널을 쓴다면 Mac에서는 터미널, Windows에서는 PowerShell을 열고 새 폴더로 이동한 뒤 `claude`를 실행합니다. 선택한 위치를 확인하고, 아래 자연어 요청을 Claude Code 대화 입력란에 넣으세요.

```text
이 연습 폴더에 data/products.json 파일을 만들어줘.
수제 디저트 상품 6개를 담은 JSON 배열로 작성해줘.
각 상품에는 id, name, price, category, image, soldOut 필드를 넣어줘.
id는 서로 다른 숫자, price는 원 단위 숫자, soldOut은 불리언으로 해줘.
품절 상품은 2개로 해줘. image는 빈 문자열로 두고, 화면에서는 이미지 없음으로 표시할 거야.
아직 화면은 만들지 말고 파일 경로와 내용을 보여줘.
같은 이름의 파일이 있으면 덮어쓰지 말고 먼저 물어봐.
```

아래는 6개 전체를 담은 예시입니다. 생성한 상품 이름이 달라도 개수·필드·자료형·품절 개수를 확인하세요. `"soldOut": true`인 초콜릿 케이크는 품절이고, `"soldOut": false`인 딸기 생크림 케이크는 판매 중입니다. 따옴표로 감싼 `"true"`·`"false"`가 아닌 불리언 값인지도 확인하세요.

```json
[
  {"id": 1, "name": "딸기 생크림 케이크", "price": 32000, "category": "케이크", "image": "", "soldOut": false},
  {"id": 2, "name": "초콜릿 케이크", "price": 35000, "category": "케이크", "image": "", "soldOut": true},
  {"id": 3, "name": "레몬 마들렌", "price": 12000, "category": "구움과자", "image": "", "soldOut": false},
  {"id": 4, "name": "버터 휘낭시에", "price": 14000, "category": "구움과자", "image": "", "soldOut": false},
  {"id": 5, "name": "바닐라 마카롱", "price": 18000, "category": "마카롱", "image": "", "soldOut": true},
  {"id": 6, "name": "말차 쿠키", "price": 10000, "category": "쿠키", "image": "", "soldOut": false}
]
```

### 2단계: Claude Code로 상품 목록 만들기

파일 확인을 마치면 같은 대화에서 다음 요청을 입력하세요.

```text
이 폴더에 Next.js로 수제 디저트 쇼핑몰의 상품 목록 화면을 만들어줘.
먼저 현재 파일과 실행 환경을 확인하고 필요한 설치 작업을 설명해줘.
상품 데이터는 data/products.json의 JSON 배열만 읽어와.
컴포넌트 안에 상품 데이터를 따로 넣거나 외부 API로 상품을 가져오지 마.
상품 6개의 이름, 원 단위 가격, 분류, 이미지 영역을 보여줘.
image가 빈 문자열이면 '이미지 없음'을 표시해줘.
id는 목록 항목의 고유 식별자로 사용하고, soldOut이 true이면 '품절'을 표시해줘.
백엔드, 데이터베이스, 주문, 결제 기능은 만들지 마.
같은 이름의 파일이 있으면 덮어쓰지 말고 먼저 물어봐.
완료하면 실행 명령과 브라우저에서 열 주소를 알려줘.
Next.js 사용을 확인할 파일과 상품 데이터를 읽는 파일의 경로도 알려줘.
```

Claude Code가 설명한 설치·실행 작업과 파일 경로를 확인한 뒤 진행합니다. 직접 실행할 명령을 안내받았다면 일반 터미널에서 해당 폴더를 열고 실행하세요. 개발 서버가 실행 중인 상태에서 안내받은 주소를 브라우저로 엽니다. 파일을 직접 여는 대신 개발 서버 주소에 접속합니다.

### 3단계: 실행한 화면과 파일 확인하기

화면을 연 뒤 다음 조건을 확인하세요. 파일 확인은 Claude Code가 알려 준 경로의 내용을 함께 읽어 봅니다.

- [ ] `package.json`의 의존성과 실행 명령에서 Next.js 사용을 확인했습니다.
- [ ] 상품을 읽는 파일이 `data/products.json`을 사용하며, 다른 상품 데이터나 외부 API를 사용하지 않습니다.
- [ ] 화면에 상품 6개가 보이고 이름·가격·분류가 목업 파일과 일치합니다. 빈 `image`는 ‘이미지 없음’으로 보이며 `id`는 고유 식별자로 사용합니다.
- [ ] `soldOut`이 `true`인 상품 2개에 ‘품절’이 표시됩니다.
- [ ] 백엔드·데이터베이스·주문·결제 기능을 만들지 않았습니다.

맞지 않는 항목이 있으면 기대한 조건과 실제 결과를 Claude Code에 알려 수정하고, 브라우저에서 다시 확인합니다. 확인한 화면을 캡처하고 조건별 확인 결과를 기록하세요. 실습을 마칠 때는 목업 파일을 읽는 상품 목록 화면이 실제로 떠 있어야 합니다.

여기까지가 이번 목업 화면 실습입니다. 실제 데이터로 바꾸는 과정이 궁금하다면 앞의 「목업 데이터에서 API로 바꾸기」와 「CS 교양: API 요청과 응답 이해하기」를 읽어 보세요. API 교체 요청은 같은 형식의 응답을 제공하는 서버가 준비된 뒤에 실행합니다.

## 과제: 내 서비스에 목업 데이터가 필요한지 판단하기

앞의 PRD 실습에서 만든 자신의 서비스 선택 근거와 PRD를 다시 엽니다. 수제 디저트 쇼핑몰 예시와 달리, 내 서비스에는 목업 데이터가 필요한 곳이 있는지 판단하고 그 근거를 적으세요. 필요하지 않다는 판단도 근거가 있으면 됩니다.

### 진행 방법

1. PRD의 핵심 화면과 요구사항을 보며, 실제 데이터를 연결하기 전에 예시 데이터로 확인할 부분이 있는지 살펴봅니다.
2. 목업 데이터가 필요한지 판단하고, 어떤 화면·요구사항을 근거로 삼았는지 `목업-데이터-판단.md`에 적습니다. 필요하다면 확인할 내용과 데이터 항목을, 필요하지 않다면 그 이유와 화면을 확인할 다른 방법을 적으세요.
3. **필요하다고 판단한 학생만 선택적으로** 목업 데이터 JSON을 만들거나 화면 구현까지 해 볼 수 있습니다. 이 경우 선택한 프레임워크, 파일 경로와 구현 범위를 Claude Code에 알려 주고, “같은 이름의 파일이 있으면 덮어쓰지 말고 먼저 물어봐”를 요청에 포함하세요. 백엔드 연결 없이 목업 파일로 화면을 확인합니다. JSON 생성과 화면 구현은 필수 제출 항목이 아닙니다.

### 제출 전 체크

- [ ] 도구의 일반적인 장점만 나열하지 않고, 내 서비스의 조건과 연결해 선택했나요?
- [ ] PRD에서 이번에 만들 범위와 나중에 만들 기능이 구별되나요?
- [ ] 목업이 필요하거나 필요하지 않다는 결론을 다른 사람이 읽어도 납득할 수 있나요?

과제까지 마치려면 시간이 더 필요하며, 읽기와 실습을 포함한 오늘 레슨 전체는 대략 4～7시간 걸립니다.

| API 연결 | 데이터 구조를 유지하면 가져오는 부분을 바꾸면서 화면을 재사용할 수 있습니다 | 정상 목록뿐 아니라 빈 목록·로딩·오류 상태도 처리합니다 |
